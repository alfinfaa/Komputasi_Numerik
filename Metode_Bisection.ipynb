# Penyelesaian Persamaan Nonlinear
## Metode Bisection (Metode Bagi Dua)

Notebook ini menunjukkan pencarian akar fungsi nonlinear dengan membagi interval yang mengapit akar menjadi dua bagian secara berulang.

## Cell 2 – Mendefinisikan fungsi
Contoh fungsi yang digunakan adalah f(x) = x³ - x - 2. Fungsi ini akan dicari nilai akarnya pada interval awal [1, 2].

```python
# Mendefinisikan fungsi nonlinear yang akan dicari akarnya
def f(x):
    return x**3 - x - 2
```

## Cell 3 – Menentukan parameter awal
Batas bawah dan batas atas membentuk interval pencarian. Toleransi menentukan ketelitian pendekatan, sedangkan iterasi maksimum membatasi banyaknya pengulangan.

```python
# Menentukan batas interval awal
a = 1.0
b = 2.0

# Menentukan toleransi dan jumlah iterasi maksimum
toleransi = 0.0001
maks_iterasi = 100

# Memeriksa apakah nilai fungsi di kedua ujung interval berbeda tanda
if f(a) * f(b) >= 0:
    raise ValueError("Interval tidak mengapit perubahan tanda fungsi.")

print("Nilai f(a):", f(a))
print("Nilai f(b):", f(b))
print("Interval awal:", a, "sampai", b)
```

## Cell 4 – Proses iterasi Bisection
Pada setiap iterasi dihitung titik tengah c. Kemudian tanda f(a) dan f(c) digunakan untuk menentukan bagian interval yang masih mengapit akar.

```python
def bisection(a, b, toleransi, maks_iterasi):
    # Menampilkan judul kolom tabel iterasi
    print("Iterasi |       a |       b |       c |      f(c)")

    # Mengulangi proses sampai toleransi tercapai atau iterasi maksimum
    for i in range(1, maks_iterasi + 1):
        # Menghitung titik tengah interval
        c = (a + b) / 2

        # Menghitung nilai fungsi pada titik tengah
        fc = f(c)

        # Menampilkan data iterasi saat ini
        print(f"{i:7d} | {a:7.5f} | {b:7.5f} | {c:7.5f} | {fc: .7f}")

        # Menghentikan proses jika nilai fungsi telah memenuhi toleransi
        if abs(fc) <= toleransi:
            return c, i

        # Memilih subinterval yang masih memiliki perubahan tanda
        if f(a) * fc < 0:
            b = c
        else:
            a = c

    # Mengembalikan pendekatan terakhir jika iterasi maksimum tercapai
    return c, maks_iterasi
```

## Cell 5 – Menjalankan metode Bisection
Fungsi dipanggil dengan interval, toleransi, dan batas iterasi yang sudah ditentukan.

```python
# Menjalankan proses pencarian akar
akar, jumlah_iterasi = bisection(a, b, toleransi, maks_iterasi)

# Menampilkan hasil akhir perhitungan
print("\nHasil perhitungan:")
print(f"Pendekatan akar = {akar:.7f}")
print("Jumlah iterasi =", jumlah_iterasi)
print(f"Nilai f(akar) = {f(akar):.10f}")
```

Berdasarkan perhitungan, metode Bisection menghasilkan pendekatan akar untuk fungsi f(x) = x³ - x - 2 pada interval [1, 2]. Ketelitian hasil mengikuti toleransi yang telah ditentukan.
