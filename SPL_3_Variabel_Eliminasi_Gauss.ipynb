# Penyelesaian Sistem Persamaan Linear Tiga Variabel
## Metode Eliminasi Gauss

Notebook ini mengikuti tahapan pada materi perkuliahan: matriks augmented, operasi baris elementer, eliminasi maju hingga bentuk segitiga atas, dan substitusi mundur.

## Cell 2 – Data persamaan
Tiga persamaan contoh yang digunakan adalah:

- 2x + y - z = 8
- -3x - y + 2z = -11
- -2x + y + 2z = -3

Data disimpan dalam bentuk matriks augmented. Kolom terakhir merupakan konstanta ruas kanan.

```python
# Matriks augmented: tiga kolom koefisien dan satu kolom konstanta
A = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

# Menampilkan matriks awal agar susunan data dapat diperiksa
print("Matriks augmented awal:")
for baris in A:
    print(baris)
```

## Cell 3 – Proses eliminasi maju (forward elimination)
Fungsi berikut mengubah matriks augmented menjadi bentuk segitiga atas. Pemilihan pivot menggunakan partial pivoting, yaitu memilih elemen dengan nilai absolut terbesar pada kolom aktif. Operasi baris ini tetap merupakan operasi baris elementer.

```python
def eliminasi_gauss(matriks):
    # Menentukan jumlah persamaan dalam matriks
    n = len(matriks)

    # Menyalin matriks agar data awal tidak ikut berubah
    M = [baris[:] for baris in matriks]

    # Memproses pivot dari kolom pertama sampai kolom terakhir koefisien
    for i in range(n):
        # Memilih baris pivot dengan nilai absolut terbesar pada kolom aktif
        baris_pivot = max(range(i, n), key=lambda r: abs(M[r][i]))

        # Menukar baris aktif dengan baris pivot terpilih
        M[i], M[baris_pivot] = M[baris_pivot], M[i]

        # Menghindari pembagian dengan pivot nol atau sangat kecil
        if abs(M[i][i]) < 1e-12:
            raise ValueError("Sistem tidak memiliki solusi tunggal.")

        # Menghilangkan elemen di bawah pivot
        for j in range(i + 1, n):
            faktor = M[j][i] / M[i][i]

            # Memperbarui koefisien dan konstanta pada baris yang dieliminasi
            for k in range(i, n + 1):
                M[j][k] = M[j][k] - faktor * M[i][k]

    # Mengembalikan matriks berbentuk segitiga atas
    return M
```

## Cell 4 – Menampilkan matriks segitiga atas
Cell ini memperlihatkan hasil eliminasi maju sebelum dilakukan substitusi mundur.

```python
matriks_segitiga = eliminasi_gauss(A)

print("Matriks setelah eliminasi Gauss:")
for baris in matriks_segitiga:
    print([round(nilai, 6) for nilai in baris])
```

## Cell 5 – Substitusi mundur (back substitution)
Karena matriks sudah berbentuk segitiga atas, nilai variabel dihitung mulai dari baris paling bawah, kemudian bergerak ke baris paling atas.

```python
def substitusi_mundur(matriks):
    # Menentukan banyaknya variabel
    n = len(matriks)

    # Menyediakan daftar untuk menyimpan nilai setiap variabel
    solusi = [0.0] * n

    # Membaca baris dari bawah ke atas
    for i in range(n - 1, -1, -1):
        # Menghitung jumlah kontribusi variabel yang sudah diketahui
        jumlah = sum(matriks[i][j] * solusi[j] for j in range(i + 1, n))

        # Menghitung nilai variabel pada baris yang sedang diproses
        solusi[i] = (matriks[i][n] - jumlah) / matriks[i][i]

    # Mengembalikan nilai variabel dalam urutan x, y, z
    return solusi
```

## Cell 6 – Menjalankan program dan menampilkan solusi
Fungsi eliminasi dan substitusi dipanggil secara berurutan. Hasil ditampilkan dengan empat angka di belakang koma.

```python
# Menghitung solusi dari matriks segitiga atas
hasil = substitusi_mundur(matriks_segitiga)

# Menampilkan solusi untuk x, y, dan z
print("Hasil penyelesaian SPL:")
print(f"x = {hasil[0]:.4f}")
print(f"y = {hasil[1]:.4f}")
print(f"z = {hasil[2]:.4f}")
```

## Cell 7 – Pemeriksaan hasil
Nilai solusi dimasukkan kembali ke persamaan awal. Jika hasil ruas kiri sama dengan konstanta pada ruas kanan, solusi telah memenuhi sistem persamaan.

```python
# Memeriksa kembali setiap persamaan menggunakan nilai hasil
persamaan = [
    2 * hasil[0] + hasil[1] - hasil[2],
    -3 * hasil[0] - hasil[1] + 2 * hasil[2],
    -2 * hasil[0] + hasil[1] + 2 * hasil[2]
]
konstanta = [8, -11, -3]

for nomor, (kiri, kanan) in enumerate(zip(persamaan, konstanta), start=1):
    print(f"Persamaan {nomor}: ruas kiri = {kiri:.4f}, ruas kanan = {kanan}")

print("Semua persamaan terpenuhi:", all(abs(kiri-kanan) < 1e-9 for kiri, kanan in zip(persamaan, konstanta)))
```

## Kesimpulan
Dari contoh perhitungan dengan metode Eliminasi Gauss diperoleh x = 2, y = 3, dan z = -1. Pemeriksaan substitusi ke persamaan awal menunjukkan bahwa ketiga persamaan terpenuhi.
